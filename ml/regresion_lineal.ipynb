```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#Datos ficticios
altura = np.array([170, 165, 180, 155, 175, 150, 185]).reshape(-1,1)
peso = np.array([65, 58, 75, 52, 68, 50, 89])

#Modelo de regresión
modelo = LinearRegression()
modelo.fit(altura, peso)

#Predicciones
predicciones = modelo.predict(altura)

#Parámetros del modelo
print("Coeficiente de intersección/Intercepto (b0):", modelo.intercept_)
print("Coeficiente de pendiente (b1):", modelo.coef_[0])
print("R^2:", r2_score(peso, predicciones))



#Gráfica
plt.scatter(altura, peso, color='blue', label='Datos reales')
plt.plot(altura, predicciones, color='red', label='Regresión lineal')
plt.xlabel('Altura (cm)')
plt.ylabel('Peso (kg)')
plt.title('Regresión lineal de altura vs. peso')
plt.legend()
plt.show()
```
